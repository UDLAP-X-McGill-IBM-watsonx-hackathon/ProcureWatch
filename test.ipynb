```python
from PIL import Image, ImageDraw, ImageFont
import textwrap

def text_to_png(text, out_path="output.png",
                width=800, font_size=32,
                font_path=None,  # e.g., "arial.ttf"
                padding=20,
                bg_color="white",
                text_color="black"):

    # Load default font or custom one
    font = ImageFont.truetype(font_path, font_size) if font_path else ImageFont.load_default()

    # Wrap text to fit the width
    wrapped = textwrap.fill(text, width=60)

    # Temporary image to calculate size
    dummy = Image.new("RGB", (width, 1))
    draw = ImageDraw.Draw(dummy)

    # Get bounding box → (left, top, right, bottom)
    left, top, right, bottom = draw.multiline_textbbox((0, 0), wrapped, font=font)

    text_width = right - left
    text_height = bottom - top

    # Create final image with padding
    img = Image.new("RGB", (text_width + padding * 2, text_height + padding * 2), bg_color)
    draw = ImageDraw.Draw(img)

    draw.multiline_text(
        (padding, padding),
        wrapped,
        fill=text_color,
        font=font
    )

    img.save(out_path)
    print(f"Saved → {out_path}")

text_to_png(response.text, "api_text.png")
```
